Tipos de Classificadores Dummy
- Mais frequente (most_frequent)
- Aleatório (uniform)
- Aleatório com distribuição igual nos dados de treino (stratifield)
- Constante e configurado pelo usuário (constant)





Tipos de Regressores Dummy
- Média dos rótulos de treinamento (mean)
- Mediana das saídas do treinamento (median)
- Quartil das saídas de treinamento (quantile)
- Constante e configurado pelo usuário (constant)

```python
from sklearn.dummy import DummyClassifier, DummyRegressor
from sklearn.datasets import load_iris, fetch_california_housing
```

```python
X, y = load_iris(return_X_y=True)
```

```python
dc = DummyClassifier(strategy='stratified')
dc.fit(X, y)
dc.score(X,y)
```

```python
X, y = fetch_california_housing(return_X_y=True)
```

```python
dr = DummyRegressor(strategy='mean')
dr.fit(X, y)
dr.score(X, y)
```
