```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Leitura dos Dados

```python
df = pd.read_csv('Mall_Customers.csv')
df.head()
```

```python
df.shape
```

Verificar dados nulos

```python
df.isnull().sum()
```

Informações Estatísticas

```python
df.describe()
```

```python
plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], marker='.')
plt.xlabel('Renda Anual [k$]')
plt.ylabel('Score (1-100)')
plt.show()
```

Selecionando dados para agrupamento

```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
X.head()
```

Importando K-means

```python
from sklearn.cluster import KMeans
```

Clusteizando com k=5

```python
modelo_kmeans = KMeans(n_clusters=5, init='k-means++')
```

```python
y_kmeans = modelo_kmeans.fit_predict(X)
```

```python
print(y_kmeans)
```

Visualizando o primeiro grupo criado

```python
print(X[y_kmeans == 0])
```

Visualizando os grupos

```python
k_grupos = 5
cores = ['r', 'b', 'k', 'y', 'g']
for k in range(k_grupos):
  cluster = X[y_kmeans == k]
  plt.scatter(cluster['Annual Income (k$)'], cluster['Spending Score (1-100)'], s=100, c=cores[k], label=f'Cluster {k}')

plt.title('Grupos de clientes')
plt.xlabel('Renda Anual (k$)')
plt.ylabel('sCORE (1-100)')
plt.grid()
plt.legend()
plt.show()
```
