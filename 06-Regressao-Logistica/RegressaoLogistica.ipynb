```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Leitura dos dados de diabetes

```python
df = pd.read_csv('diabetes.csv')
```

```python
df.head()
```

Informações estatísticas dos dados

```python
df.describe()
```

Gerando Gráfico

```python
plt.plot(df['Age'], df['Pregnancies'], 'o')
plt.xlabel('Idade')
plt.ylabel('Gravidez')
plt.show()
```

Separação de váriavel dependente

```python
y = df['Outcome']
```

```python
x = df.drop('Outcome', axis = 1)
x.head()
```

Separar os dados de treino e teste

```python
from sklearn.model_selection import train_test_split
```

```python
x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.2)
```

Selecionar modelos de regressão logistica

```python
from sklearn.linear_model import LogisticRegression
```

```python
modelo = LogisticRegression(max_iter=5000)
```

Treinamento do modelo

```python
modelo.fit(x_treino, y_treino)
```

Realizando as previsões

```python
y_previsto = modelo.predict(x_teste)
```

```python
print(y_previsto[:10])
```

Validando modelo

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_teste, y_previsto)
```

```python
print(accuracy*100)
```
